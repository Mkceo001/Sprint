## CONSTRAINTS

Constraints are rules applied to database columns.
    
Common constraints:
PRIMARY KEY,
FOREIGN KEY,
NOT NULL,
UNIQUE,
DEFAULT,
CHECK.

NOTNULL - The column cannot be empty.-name VARCHAR(100) NOT NULL

UNIQUE - Values cannot be duplicated.- email VARCHAR(100) UNIQUE

DEFAULT - Provides a default value.- status VARCHAR(20) DEFAULT 'Pending'

CHECK - Validates a condition.- price = -500

#### Example:

CREATE TABLE Students ( Student_ID INT PRIMARY KEY, Name VARCHAR(50) NOT NULL, Age INT CHECK (Age >= 18), City VARCHAR(30) DEFAULT 'Chennai' );

In this example,

PRIMARY KEY makes Student_ID unique.
NOT NULL ensures Name cannot be empty.
CHECK ensures Age is at least 18.
DEFAULT sets City to "Chennai" if no value is given.

## HAVING

HAVING filters groups.

### Example:

Find customers whose total spending is greater than ₹5,000.

SELECT

    customer_id,
    
    SUM(amount) AS total_spending
    
FROM orders

GROUP BY customer_id

HAVING SUM(amount) > 5000;

+

## RIGHT JOIN
Opposite concept of LEFT JOIN.
### Example
SELECT

    customers.name,
    
    orders.order_id,
    
    orders.amount
    
FROM customers

RIGHT JOIN orders

ON customers.customer_id = orders.customer_id;

MySQL supports RIGHT JOIN; some database systems provide equivalent results by reversing the tables and using LEFT JOIN.

## FULL JOIN

## WINDOW FUNCTIONS
Window functions are one of the most important advanced SQL concepts.
They perform calculations across related rows without collapsing those rows into one row.
This is the major difference from GROUP BY.

### ROW_NUMBER()
Assigns a unique sequential number to rows.
| employee | department | salary |
|---|---|---:|
| Arun | IT | 60000 |
| Kumar | IT | 50000 |
| Priya | HR | 55000 |
| Divya | HR | 45000 |
#### Example:
SELECT

    employee,
    
    department,
    
    salary,
    
    ROW_NUMBER() OVER (
    
        PARTITION BY department
        
        ORDER BY salary DESC
        
    ) AS row_num
    
FROM employees;
| employee | department | salary | row_num |
|---|---|---:|---:|
| Arun | IT | 60000 | 1 |
| Kumar | IT | 50000 | 2 |
| Priya | HR | 55000 | 1 |
| Divya | HR | 45000 | 2 |

### RANK()
RANK() gives the same rank to tied values.
#### Example:
| Employee | Salary |
|---|---:|
| Arun | 60000 |
| Kumar | 60000 |
| Priya | 50000 |
SELECT

    employee,
    
    salary,
    
    RANK() OVER (
    
        ORDER BY salary DESC
        
    ) AS salary_rank
    
FROM employees;
| employee | salary | rank |
|---|---:|---:|
| Arun | 60000 | 1 |
| Kumar | 60000 | 1 |
| Priya | 50000 | 3 |

### DENSE_RANK()
Similar to RANK(), but doesn't skip ranking numbers
#### Example:

SELECT

    employee,
    
    salary,
    
    DENSE_RANK() OVER (
    
        ORDER BY salary DESC
        
    ) AS salary_rank
    
FROM employees;
| employee | salary | rank |
|---|---:|---:|
| Arun | 60000 | 1 |
| Kumar | 60000 | 1 |
| Priya | 50000 | 2 |

### LAG()
LAG() gets a value from a previous row.
#### Example:
| month | sales |
|---|---:|
| January | 10000 |
| February | 15000 |
| March | 12000 |

SELECT

    month,
    
    sales,
    
    LAG(sales) OVER (
    
        ORDER BY month
        
    ) AS previous_month_sales
    
FROM monthly_sales;
| month | sales | previous_month_sales |
|---|---:|---:|
| January | 10000 | NULL |
| February | 15000 | 10000 |
| March | 12000 | 15000 |


### LEAD()
LEAD() gets a value from the next row.
#### Example:

SELECT

    month,
    
    sales,
    
    LEAD(sales) OVER (
    
        ORDER BY month
        
    ) AS next_month_sales
    
FROM monthly_sales;
| month | sales | next_month_sales |
|---|---:|---:|
| January | 10000 | 15000 |
| February | 15000 | 12000 |
| March | 12000 | NULL |
