## Reading Text Files

Reading a text file means retrieving the data stored in a file without changing it.

 Method:

 read()
 
 readline()
 
 readlines()
 

#### read()

```python
with open("Students.txt", "r") as file:
    data = file.read()

print(data)
```

#### readline()

```python
with open("students.txt", "r") as file:
    line = file.readline()

print(line)
```

#### readlines()

```python
with open("students.txt", "r") as file:
    lines = file.readlines()

print(lines)
```

## Writing CSV

```python
import csv

employees = [
    ["id", "name", "salary"],
    [101, "Mugesh", 30000],
    [102, "Arun", 35000],
    [103, "Priya", 40000]
]

with open("employees.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerows(employees)
```

## JSON File


JSON means
JavaScript Object Notation
It is commonly used for storing structured information.

```python
{
    "name": "Mugesh",
    "age": 21,
    "skills": [
        "Python",
        "SQL",
        "Machine Learning"
    ]
}
```

### Writing JSON:

```python
import json

student = {
    "name": "Mugesh",
    "age": 21,
    "skills": ["Python", "SQL", "Machine Learning"]
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)
```

```python
import json

with open("student.json", "r") as file:
    student = json.load(file)

print(student)
```

## Context Manager — with

This is one of the most important concepts in file handling.It automatically closes the file after the with block.


### Example:

```python
with open("sample1.txt", "w") as file:
    file.write("Hello World")

print("File created successfully.")
```

```python
with open("sample1.txt", "r") as file:
    data = file.read()

print(data)
```

## File Validation

Before processing a file, applications should validate it.

Validate File Extension

Validate File Size

```python
from pathlib import Path

def validate_file(file_path):

    path = Path(file_path)

    if not path.exists():
        return "File does not exist"

    if not path.is_file():
        return "Path is not a file"

    if path.stat().st_size == 0:
        return "File is empty"

    return "File is valid"


print(validate_file("dataset.csv"))
```
