## Constructor (__init__)

### what isn constructor (__init__) ?

A Constructor is a special method in Python that is automatically executed when an object is created.
It is commonly used to initialize the object's data.
The constructor eliminates the need to assign values separately after creating an object.

### Example:

When creating a student account, the student's name and age can be initialized automatically.

```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def vote(self):
        if self.age >= 18:
            print(self.name, "is Eligible to Vote")
        else:
            print(self.name, "is Not Eligible to Vote")


p1 = Person("Rahul", 17)
p1.vote()
```

## Instance Variables

Instance variables belong to a specific object.
Different objects can have different values for the same variable.

### Example 

Two students can have different names and marks.

```python
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Mugesh")
s2 = Student("Rahul")

print(s1.name)
print(s2.name)
```

## Class Variables

A class variable is shared by all objects of the class.
It is useful when the same information applies to every object.
Using class variables helps avoid duplication of shared data and ensures consistency across all objects created from the same class.

### Example:

All students may belong to the same college.

```python
class Student:
    college = "ABC College"

s1 = Student()
s2 = Student()

print(s1.college)
print(s2.college)
```

## Instance Methods

Instance Methods are methods that operate on individual objects of a class. They are defined with the self parameter,

### Example:

A bank account can have a deposit() method that changes the balance of one account.

```python
class BankAccount:
    def deposit(self, amount):
        print("Deposited:", amount)

account = BankAccount()
account.deposit(5000)
```

## Static Methods

A static method does not depend on object or class data.
It is usually used for a utility operation related to the class.
They are defined using the @staticmethod decorator and do not require the self or cls parameters.

### Example :

A calculator class can have a method to check whether a number is even.

```python
class Calculator:
    @staticmethod
    def is_even(number):
        return number % 2 == 0

print(Calculator.is_even(10))
```

## Class Methods

Class Methods are methods that operate on the class itself rather than on individual objects. They are defined using the @classmethod decorator and receive the class as their first parameter, conventionally named cls.

### Example :

A company can maintain a common employee count for all employees.

```python
class Employee:
    company = "Stackly"

    @classmethod
    def show_company(cls):
        print(cls.company)

Employee.show_company()
```

## Properties

A Property in Python is a feature that allows a method to be accessed like an attribute using the @property decorator
It is useful when you want controlled access to an object's data

### Example:

A bank account can calculate the balance through a property.

```python
class Account:
    def __init__(self, balance):
        self._balance = balance

    @property
    def balance(self):
        return self._balance

account = Account(10000)
print(account.balance)
```

## Encapsulation

Encapsulation means combining data(attributes) and methods (Function) inside a class and controlling access to the data.
It helps protect important data from unwanted direct modification.

### Example :

A bank account should not allow anyone to directly change the balance.

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance      

    def deposit(self, amount):
        self.__balance += amount

    def show_balance(self):
        print("Balance =", self.__balance)


b1 = BankAccount(5000)

b1.deposit(1057)
b1.show_balance()
```

## Inheritance

Inheritance allows one class to reuse properties and methods from another class.
It reduces duplicate code and creates a parent class And child class relationship.

### Example:

A Dog can inherit common features from an Animal.

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

dog = Dog()
dog.eat()
dog.bark()
```

## Multiple Inheritance

Multiple inheritance means one child class inherits from more than one parent class.
The child can use features from both parent classes.

### Example:


A smart phone can have features of both a Camera and a Phone.

```python
class Camera:
    def take_photo(self):
        print("Photo taken")

class Phone:
    def call(self):
        print("Calling")

class Smartphone(Camera, Phone):
    pass

s = Smartphone()
s.take_photo()
s.call()
```

## Multilevel Inheritance

Multilevel inheritance creates a chain of inheritance.
One class inherits from another class, which itself inherits from another class.

### Example

Animal → Dog → Puppy.

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

class Puppy(Dog):
    def play(self):
        print("Playing")

p = Puppy()
p.eat()
p.bark()
p.play()
```

## Hierarchical Inheritance

Hierarchical inheritance means multiple child classes inherit from one parent class.
The child classes share common functionality but can have their own behavior.

### Example:

Vehicle can be the parent of Car, Bike, and Bus.

```python
class Vehicle:
    def start(self):
        print("Vehicle started")

class Car(Vehicle):
    pass

class Bike(Vehicle):
    pass
```

## Polymorphism

Polymorphism means one interface or method name can behave differently for different objects.
It allows the same operation to work with different types of objects.

### Example:

Different payment methods can all use pay(), but each processes payment differently.

```python
class UPI:
    def pay(self):
        print("Paid using UPI")

class Card:
    def pay(self):
        print("Paid using Card")

for payment in [UPI(), Card()]:
    payment.pay()
```

## Method Overriding


child class provides its own implementation of a method that is already defined in its parent class. The child class uses the same method name, the same parameters, and the same return type (where applicable), but changes the behavior according to its own requirements.

### Example:

Different employees can have different implementations of work.

```python
class Employee:
    def work(self):
        print("Employee is working")

class Developer(Employee):
    def work(self):
        print("Developer is writing code")

d = Developer()
d.work()
```

## Method Overloading

Traditional method overloading with multiple methods having the same name is not directly supported in Python.
Instead, Python commonly uses default arguments, *args, or **kwargs.

### Example:

A calculator can perform addition with either two or three numbers.A calculator can perform addition with either two or three numbers.

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

print(Calculator().add(10, 20))
print(Calculator().add(10, 20, 30))
```

## Composition

Composition means one class contains an object of another class and strongly depends on it.
It represents a "has-a" relationship.

### Example:

A Car has an Engine. The engine is considered part of the car.

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()

car = Car()
car.engine.start()
```

## Aggregation

Aggregation is also a "has-a" relationship, but the contained object can exist independently.
The parent object does not completely control the lifetime of the child object.

### Example

A Department has employees, but an employee can exist even if the department is removed.

```python
class Employee:
    def __init__(self, name):
        self.name = name

class Department:
    def __init__(self, employee):
        self.employee = employee

emp = Employee("Mugesh")
dept = Department(emp)
```

## Association

Association means two independent classes are connected or interact with each other.
Neither object necessarily owns the other.

### Example:

A Teacher teaches a Student, but both can exist independently.

```python
class Teacher:
    def teach(self, student):
        print("Teaching", student.name)

class Student:
    def __init__(self, name):
        self.name = name

student = Student("Mugesh")
teacher = Teacher()

teacher.teach(student)
```

## super()

super() is used to access methods or the constructor of a parent class.
It is especially useful when extending parent functionality in a child class.

### Example:

A child employee class can call the parent constructor and then add its own information

```python
class Employee:
    def __init__(self, name):
        self.name = name

class Developer(Employee):
    def __init__(self, name, language):
        super().__init__(name)
        self.language = language

d = Developer("Mugesh", "Python")
```

##  __str__()

__str__() defines the human-readable representation of an object.
It is automatically called when you use print(object).

### Example:

Instead of printing a memory address for an employee object, you can display the employee's name.

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name

e = Employee("Mugesh")
print(e)
```

## __repr__()

__repr__() provides a more detailed representation of an object, mainly useful for developers and debugging.
It should ideally contain enough information to understand the object's state.

### Example:

When debugging an employee object, you may want to see both the name and employee ID.

```python
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def __repr__(self):
        return f"Employee({self.name!r}, {self.emp_id!r})"

e = Employee("Mugesh", 101)
print(repr(e))
```

## __len__()

The__len__() defines what should happen when Python's len() function is used on your object.
It allows your custom object to behave like a collectioTn.

### Example:

A shopping cart can return the number of products inside it.

```python
class Cart:
    def __init__(self):
        self.items = ["Laptop", "Mouse", "Keyboard"]

    def __len__(self):
        return len(self.items)

cart = Cart()
print(len(cart))
```

## Dataclasses

A dataclass makes it easier to create classes mainly used for storing data.
Python automatically provides methods such as __init__() and a useful representation.

### Example:

A user profile containing name, age, and email can be represented using a dataclass.

```python
from dataclasses import dataclass

@dataclass
class User:
    name: str
    age: int
    email: str

user = User("Mugesh", 21, "mugesh@example.com")
print(user)
```

## Singleton Pattern

The Singleton pattern is designed so that a class has only one shared instance.
It can be useful when multiple parts of an application need to access the same shared resource.

### Example:

An application might use one shared configuration manager.

```python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

a = Singleton()
b = Singleton()

print(a is b)
```
