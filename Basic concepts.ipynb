List comprehension :

List comprehension is a short and simple way to create a new list using a loop.

we want to create a new list quickly from an existing sequence. when you need to create, filter, or modify a list in a short way.

For Example:

Imagine you have a list of numbers and want to get only the even numbers.

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [num for num in numbers if num % 2 == 0]

print(even_numbers)
```


Dictionary Comprehension:

Example:

Imagine you have a list of students and want to give each student a mark of 80

```python
students = ["Mugesh", "Arun", "Rahul"]

marks = {student: 80 for student in students}

print(marks)
```

Variable-Length Arguments (*args, **kwargs):

Sometimes, when creating a function, we don't know how many values the user will give.

For example, a function may receive 2 numbers today, but 5 numbers tomorrow.

*args and **kwargs are useful.

*args:

*args allows a function to accept any number of normal/positional values.

Example:

Imagine a teacher asks:
"Tell me the names of all the students in your group."

The group could have 2, 5, or 10 students.

```python
def students(*names):
    print(names)

students("Mugesh", "Arun", "Rahul")
```

We can give more input :

**kwargs :

**kwargs — Multiple Keyword Values

**kwargs allows a function to accept any number of key=value arguments.

```python
def student_info(**details):
    print(details)

student_info(name="Mugesh", age=21, city="Chennai")
#u can add more unique values
student_info(
    name="Mugesh",
    age=21,
    city="Chennai",
    course="AI/ML"
)
```

Recursive function:

A recursive function is a function that calls itself until a condition tells it to stop.

Without a stopping condition, the function could keep calling itself indefinitely.

Example:

Simple Real-Life Example

Imagine you are counting down:

5  4  3  2  1  Stop

The same action is repeated until we reach the stopping point.

```python
def countdown(n):
    if n == 0:
        print("Done!")
    else:
        print(n)
        countdown(n - 1)

countdown(5)
```
