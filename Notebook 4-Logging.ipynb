# Logging

## What Is Logging ?

Logging is the process of recording information about what happens while a program is running.

Think of logging as a diary for your application. Your program writes down what happened, when it happened, and whether something went wrong

### Example:

magine an online shopping application:

10:30:01 - User logged in

10:30:05 - Product searched

10:30:10 - Payment started

10:30:12 - Payment successful

10:30:15 - Order created

These records are called logs.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("User logged in")
logging.warning("Password attempt is high")
logging.error("Payment failed")
```

## Why Use Logging?

In a Small Program we use print() but in a real application log is best to control.

Logging helps us:
- Monitor applications
- Find errors
- Debug problems
- Track user activities
- Monitor APIs
- Monitor ML model predictions
- Save errors to files
- Separate important messages from normal messages
- Investigate problems after they happen

## print() vs logging

### Print()

Its a simple and mostly used for displaying output and its has Difficult to manage in large applications.

### Example


```python
def login(username):
    print("Login process started")
    print(f"User {username} is trying to login")

    print(f"User {username} logged in successfully")


login("Mugesh")
```

### logging

It is an more powerfull and it Designed for application monitoring and It easy to manage.

### Example:


```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

def login(username):
    logging.info("Login process started")
    logging.info(f"User {username} is trying to login")

    logging.info(f"User {username} logged in successfully")


login("Mugesh")
```

## Log Levels

Logging provides different levels to indicate the importance of a message.

DEBUG

INFO

WARNING

ERROR

CRITICAL

### DEBUG

DEBUG is used for detailed information useful during development and troubleshooting.These details can help developers understand exactly what the application is doing.

#### Example:


Imagine an e-commerce application:

```python
logging.debug("Database connection object created")
logging.debug("User ID = 105")
logging.debug("Cart contains 3 products")
logging.debug("Total amount = 1500")
```

### INFO

INFO represents normal application activity.

#### Example:

```python
logging.info("User logged in")
logging.info("Order created successfully")
logging.info("Model loaded successfully")
```

### WARNING

WARNING means something unexpected happened, but the application can still continue.

#### Example :

```python
logging.warning("Disk space is running low")
```

### Error 

ERROR means something went wrong and a particular operation failed.

#### Example:

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    logging.error("Cannot divide by zero")
```

### CRITICAL

CRITICAL represents a very serious problem

#### Example

```python
logging.critical("Database server is completely unavailable")
```

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.debug("Debug message")
logging.info("Info message")
logging.warning("Warning message")
logging.error("Error message")
logging.critical("Critical message")
```

## Log Formatters

A formatter controls how a log message looks.

### Example

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("User logged in")
```

2026-09-24 11:30:20,123 - INFO - User logged in

### Important Formatter Components

%(asctime)s - Shows the time - 2026-09-24 11:30:20

%(levelname)s - Shows the log level.

INFO

ERROR

WARNING

%(message)s - Shows your actual message. - User logged in

(name)s - Shows the logger name.

%(filename)s - Shows the Python file name.

%(lineno)d - Shows the line number where the log was generated

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(filename)s:%(lineno)d | %(message)s"
)

logging.info("Application started")
```

## File Handlers

A FileHandler allows us to save logs into a file.

### Example:

```python
import logging

logger = logging.getLogger(__name__)

logger.setLevel(logging.INFO)

file_handler = logging.FileHandler("application.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

file_handler.setFormatter(formatter)

logger.addHandler(file_handler)

logger.info("Application started")
logger.error("Something went wrong")
```

## Console Handlers

A ConsoleHandler sends logs to the console/terminal.

```python
import logging

logger = logging.getLogger(__name__)

logger.setLevel(logging.INFO)

console_handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)

logger.addHandler(console_handler)

logger.info("Server started")
```

## Rotating Logs

Imagine your application runs continuously for one year.
Rotating logs automatically creates new log files when the old file reaches a certain size.Your log file could become:
app.log

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

username = "mugesh"

logging.info(f"Login attempt by user: {username}")

password_correct = True

if password_correct:
    logging.info(f"User {username} logged in successfully")
else:
    logging.warning(f"Incorrect password for user {username}")
```

## Logging Best Practices

 Use meaning message - logging.info("Customer order  created successfully")

 Use the correct log level - logging.info("User logged in")

 Don't log sensitive information - logging.info("User authentication attempted")

 Don't log everything - 

 Use DEBUG for detailed information  - logging.info(f"Variable x = {x}")

 Include useful context - logging.error("Failed to load model: model file not found")

 Use timestamps  - format="%(asctime)s - %(levelname)s - %(message)s"

 Save important logs - Console + Log file

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger(__name__)
logger.setLevel(logging.DEBUG)

formatter = logging.Formatter(
    "%(asctime)s | %(levelname)s | %(name)s | %(message)s"
)

# Console Handler
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)
console_handler.setFormatter(formatter)

# File Handler
file_handler = RotatingFileHandler(
    "application.log",
    maxBytes=1024 * 1024,
    backupCount=3
)

file_handler.setLevel(logging.DEBUG)
file_handler.setFormatter(formatter)

# Add handlers
logger.addHandler(console_handler)
logger.addHandler(file_handler)


def login(username, password):
    logger.info("Login attempt for user: %s", username)

    if password == "admin123":
        logger.info("Login successful for user: %s", username)
        return True

    logger.warning("Login failed for user: %s", username)
    return False


try:
    logger.info("Application started")

    login("mugesh", "admin123")

except Exception:
    logger.exception("Unexpected application error")  
```
