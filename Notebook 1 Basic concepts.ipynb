Recursive function:

A recursive function is a function that calls itself until a condition tells it to stop.

Without a stopping condition, the function could keep calling itself indefinitely.

Example:

Simple Real-Life Example

Imagine you are counting down:

5  4  3  2  1  Stop

The same action is repeated until we reach the stopping point.

```python
def countdown(n):
    if n == 0:
        print("Done!")
    else:
        print(n)
        countdown(n - 1)

countdown(5)
```

## Generators

A generator is a special type of iterator.
It generates values one at a time, instead of storing all values in memory.


### Example :

Imagine you need a number of 1 to 1 crore.
It produce One number at a time 


```python
def count():
    for i in range(1, 6):
        yield i
```

```python
for number in count():
    print(number)
```

## Yield 

Yield is mainly used inside the generator.
It works somewhat like return,It not end the function it woould be pauses the function.

### Example :

Suppose a restaurant has 5 orders. You don't want to load all orders into memory at once. You can process and serve them one by one using yield.

```python
def process_orders():
    orders = ["Pizza", "Burger", "Biryani", "Dosa", "Pasta"]

    for order in orders:
        print("Preparing:", order)
        yield order

orders = process_orders()

print(next(orders))
print(next(orders))
print(next(orders))
```

## Decorator

A decorator is a function that adds extra functionality to another function without changing the original function's code.

### Example:

Think of it like a gift wrapper :

The gift = your original function

The wrapper = decorator

Extra decoration = additional functionality

```python
def decorator(function):
    def wrapper():
        print("Before function")
        function()
        print("After function")
    return wrapper


@decorator
def hello():
    print("Hello Mugesh!")


hello()
```

## Closures

A closure is when an inner function remembers a variable from the outer function even after the outer function has finished.
Closures are commonly used with decorators, because decorators use an inner wrapper() function that remembers the original function.

### Example :

Imagine you have a bank account.
    you create a account their have a balanced. you have a to function like deposit and withdraw to remember the balanced.

```python
def bank_account(balance):

    def deposit(amount):
        nonlocal balance
        balance += amount
        return balance

    return deposit


account = bank_account(10000)

print(account(2000))
print(account(3000))
```

### Global

Global is used to change the variable in whole program.

Imagine a bank has a variable storing its total money.

## Example :

Bank total money = ₹10,00,000
A function called deposit needs to increase that amount.
The money belongs to the whole program, not specifically to the function.

```python
balance = 10000

def deposit():
    global balance
    balance += 2000

deposit()

print(balance)
```

### Nonlocal

nonlocal is want to change a variable from the outer function.it used in nested function.
nonlocal count allows the inner function to change the variable from the outer function.

```python
def outer():
    count = 0

    def increase():
        nonlocal count
        count += 1
        print(count)

    increase()

outer()
```

## Built-in Functions

## map()

### What is map() ?

map() applies the same operation to every item in a collection.

### Example:

Imagine a teacher wants to give +5 marks to every student's score.

```python
marks = [50, 60, 70, 80]

new_marks = map(lambda x: x + 5, marks)

print(list(new_marks))
```

### filter()

### What does filter() do?

filter() selects only the items that satisfy a condition.

### Example :

Imagine a teacher wants to select students who scored more than 50.

```python
marks = [35, 60, 45, 80, 90]

result = filter(lambda x: x > 50, marks)

print(list(result))
```
