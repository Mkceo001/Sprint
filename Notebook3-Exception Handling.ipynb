# Exception Handling

An exception is an unexpected problem or error that occurs while a Python program is running and interrupts the normal flow of execution. Exceptions can happen for many reasons, such as entering invalid input, trying to divide a number by zero, accessing a file that does not exist, using an incorrect list index, or importing a module that is unavailable. If an exception is not handled, Python stops the program and displays an error message. Exception handling allows us to detect these errors and respond to them properly instead of allowing the entire program to crash. Python provides keywords such as try, except, else, finally, and raise to handle exceptions. In real-world applications such as banking systems, login applications, e-commerce websites, database applications, and AI/ML projects, exception handling is important because it makes programs more reliable, stable, user-friendly, and easier to debug.

## try

The try block contains code that might cause an exception.

Usually, try is used together with except.

## Except

except is used to catch and handle an exception.

### Example:

```python
try:
    number = int(input("Enter a number: "))
    print(number)

except ValueError:
    print("Please enter a valid number.")
```

If you enter a value as numerical is correct and other values like string  as get error and give output as invalid.

## Else

The else block executes only when the try block succeeds without an exception.

### Example:

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid number.")

else:
     print("You entered:", number)
```

## Finally

finally always executes, whether an exception occurs or not.

### Example

Closing files

Closing database connections

Releasing resources

Logging completion

Closing network connections

```python
try:
    number = int(input("Enter a number: "))
    print(number)

except ValueError:
    print("Invalid input.")

finally:
    print("Program execution completed.")
```

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid number.")

else:
    print("Number entered successfully:", number)

finally:
    print("Thank you!")
```

## Raise

raise allows you to manually generate an exception.
Python automatically generates many exceptions.
But sometimes your application needs to decide when something should be considered an error.

### Example:

```python
balance = 5000
amount = 7000

try:
    if amount > balance:
        raise ValueError("Insufficient balance")

    balance -= amount
    print("Withdrawal successful")

except ValueError as e:
    print("Transaction failed:", e)
```

We created the exception ourselves using raise.

## Assert

assert is used to check whether a condition is true.

### Example:

```python
age = int(input("Enter a Age: "))
assert age >= 18

print("Eligible")
```

## Custom Exceptions

Python already provides many built-in exceptions:

ValueError

TypeError

IndexError

KeyError

But real applications sometimes need their own meaningful exception types.
    
These are called custom exceptions.

### Example :

```python
class InsufficientBalanceError(Exception):
    pass
balance = 5000
amount = 7000

try:
    if amount > balance:
        raise InsufficientBalanceError("Insufficient account balance")

    balance -= amount
    print("Withdrawal successful")

except InsufficientBalanceError as e:
    print("Transaction failed:", e)
```

### ValueError

ValueError occurs when a function receives a value of the correct general type, but the value itself is inappropriate.
If there is an int datatype but the use give input as string iscause value error.

### Example:

```python
try:
    age = int(input("Enter your age: "))
    print("Age:", age)

except ValueError:
    print("Please enter numbers only.")
```

## Type Error

TypeError occurs when you perform an operation using an inappropriate data type.

### Example:

```python
try:
    age = 20
    name = "Mugesh"

    result = age + name

except TypeError:
    print("Cannot add integer and string.")
```

## IndexError


IndexError occurs when you try to access a list, tuple, or similar sequence using an index that doesn't exist.

### Example:

Imagine a college application displaying students:

```python
students = ["Mugesh", "Arun", "Kumar"]

try:
    print(students[5])

except IndexError:
    print("Student index does not exist.")
```

## Key Error

KeyError usually occurs when you try to access a dictionary key that doesn't exist.

### Example:

```python
student = {
    "name": "Mugesh",
    "age": 21
}

try:
    print(student["salary"])

except KeyError:
    print("Salary information is not available.")
```

## AttributeError

AttributeError occurs when you try to access an attribute or method that an object doesn't have.

### Example:

```python
name = "Mugesh"

try:
    name.calculate_salary()

except AttributeError:
    print("This object does not have that method.")
```

## Name Error

NameError occurs when Python cannot find a variable or name that you're trying to use.

### Example:

```python
try:
    print(username)

except NameError:
    print("The variable has not been defined.")
```

## FileNotFoundError

It occurs when it tries to open a file that doesn't exist at the specified path.

### Example:

```python
try:
    file = open("students.txt", "r")
    data = file.read()
    file.close()

except FileNotFoundError:
    print("The file was not found.")
```

## ZeroDivisionError

It occurs when you divide a number by zero.

### Example:

```python
try:
    result = 100 / 0
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

## ImportError

ImportError occurs when Program cannot import something that an import statement expects to import.

### Example:

```python
try:
    from math import hello

except ImportError:
    print("The requested item could not be imported.")
```

## ModuleNotFoundError

ModuleNotFoundError occurs when Program cannot find the module you're trying to import.

### Example:

```python
try:
    import mk

except ModuleNotFoundError:
    print("The required module is not installed.")
```

```python
class InsufficientBalanceError(Exception):
    pass


balance = 5000

try:
    amount = int(input("Enter withdrawal amount: "))

    if amount <= 0:
        raise ValueError("Amount must be greater than zero.")

    if amount > balance:
        raise InsufficientBalanceError("Insufficient balance.")

    balance -= amount

except ValueError as e:
    print("Invalid amount:", e)

except InsufficientBalanceError as e:
    print("Transaction failed:", e)

else:
    print("Withdrawal successful.")
    print("Remaining balance:", balance)

finally:
    print("Thank you for using our banking system.")
```
